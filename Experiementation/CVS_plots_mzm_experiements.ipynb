# 16 QAM No DPD and DPD trained with BPS 100kHz at 0.5 mzm scale

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load both result files
df_no = pd.read_csv("results_No_DPD_16qam_100_mzmchange.csv")
df_dpd = pd.read_csv("results_16qam_100_0.5_DPD_bps.csv")

# Add a label so we know which data is which
df_no["DPD"] = "No DPD"
df_dpd["DPD"] = "DPD_bps"

# Merge into one DataFrame
df = pd.concat([df_no, df_dpd], ignore_index=True)
```

```python
plt.figure(figsize=(10,6))

for dpd_option in df["DPD"].unique():              # "No DPD", "DPD"
    for da in [True, False]:                       # True = DA, False = NDA
        for cpr_mode in df["CPR"].unique():        # "bps", "ddpll"
            
            subset = df[(df["DPD"] == dpd_option) &
                        (df["Data_Aided"] == da) &
                        (df["CPR"] == cpr_mode)]
            
            if len(subset) == 0:
                continue
            
            label = f"{dpd_option} | {'DA' if da else 'NDA'} | {cpr_mode}"
            
            plt.semilogy(
                subset["mzmScale"], 
                subset["BER"], 
                marker='o', 
                linestyle='-', 
                label=label
            )

plt.xlabel("MZM Scale")
plt.ylabel("BER")
plt.title("BER vs MZM Scale (DPD vs No-DPD, DA and NDA)")
plt.grid(True, which='both', ls='--')
plt.legend(fontsize=8)
plt.tight_layout()
plt.show()
```

# Comparison of NO DPD 64QAM 10kHz and 100kHZ

# 16 QAM No DPD, DPD trained with BPS 10kHz at 0.5 mzm scale, DPD trained with DDPL 10kHz at 0.5 mzm scale just to see the difference

```python
import matplotlib.pyplot as plt
import pandas as pd

# -------------------------------------------------
# 1. Load your 3 CSV files
# -------------------------------------------------
df1 = pd.read_csv("results_No_DPD_16qam_100_mzmchange.csv")
df2 = pd.read_csv("results_16qam_100_0.5_DPD_bps.csv")
df3 = pd.read_csv("results_DPD_16_ddpll_100_0.5.csv")

# -------------------------------------------------
# 2. Assign DPD labels
# -------------------------------------------------
df1["DPD"] = "No DPD"
df2["DPD"] = "DPD BPS"
df3["DPD"] = "DPD DDPLL"

# Combine into a single dataframe
df = pd.concat([df1, df2, df3], ignore_index=True)

# -------------------------------------------------
# 3. PLOTTING (same structure as your code)
# -------------------------------------------------
plt.figure(figsize=(10,6))

for dpd_option in df["DPD"].unique():              # "No DPD", "DPD 1", "DPD 2"
    for da in [True, False]:                       # True = DA, False = NDA
        for cpr_mode in df["CPR"].unique():        # "bps", "ddpll"
            
            subset = df[(df["DPD"] == dpd_option) &
                          (df["Data_Aided"] == da) &
                          (df["CPR"] == cpr_mode)]
            
            if len(subset) == 0:
                continue
            
            # Sort for clean lines
            subset = subset.sort_values("mzmScale")

            label = f"{dpd_option} | {'DA' if da else 'NDA'} | {cpr_mode}"
            
            plt.semilogy(
                subset["mzmScale"], 
                subset["BER"], 
                marker='o', 
                linestyle='-', 
                label=label
            )

plt.xlabel("MZM Scale")
plt.ylabel("BER")
plt.title("BER vs MZM Scale (3 datasets: DPD BPS, DPD DDPLL, No-DPD, DA/NDA, CPR modes)")
plt.grid(True, which='both', ls='--')
plt.legend(fontsize=8)
plt.tight_layout()
plt.show()
```

# No DPD L Span 16qam 100 0.5 doing 8,4,2,1 amplifiers

```python
import matplotlib.pyplot as plt
import pandas as pd

# Read the CSV file directly (no use of 'results')
df = pd.read_csv("results_No_DPD_lspan_experiment_16qam.csv")

# Divide lspan by 10
df["lspan"] = 80/df["lspan"]

# Two subplots: DA (left), NDA (right)
fig, axes = plt.subplots(1, 2, figsize=(14, 5), sharey=False)

subplot_map = {True: 0, False: 1}

for da in [True, False]:
    ax = axes[subplot_map[da]]

    for cpr_mode in df["CPR"].unique():

        subset = df[(df["Data_Aided"] == da) &
                    (df["CPR"] == cpr_mode)]

        if len(subset) == 0:
            continue

        # Sort so the plot doesn't zig-zag
        subset = subset.sort_values("lspan")

        label = f"{cpr_mode}"

        ax.semilogy(
            subset["lspan"],
            subset["BER"],
            marker="o",
            linestyle="-",
            label=label
        )

    # Titles
    ax.set_title("DA" if da else "NDA")

    # X label
    ax.set_xlabel("Number of Spans")
    ax.grid(True, which='both', ls='--')

    # Invert x-axis (largest span → smallest)
    #ax.invert_xaxis()

    # Separate Y-axis titles
    if da:
        ax.set_ylabel("BER (zoomed)")
    else:
        ax.set_ylabel("BER (NDA scale)")

    # Zoom DA Y-axis
    if da:
        da_vals = df[df["Data_Aided"] == True]["BER"]
        ymin = da_vals.min() * 0.8
        ymax = da_vals.max() * 1.2
        ax.set_ylim([ymin, ymax])

# Put legend only on NDA subplot
axes[0].legend(title="CPR Mode")
axes[1].legend(title="CPR Mode")

plt.suptitle("BER vs lspan (DA zoomed, NDA separate y-axis)", fontsize=14)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# ---- CSV FILES TO COMPARE ----
file1 = "results_No_DPD_lspan_experiment_16qam.csv"
file2 = "results_DPD_lspan_16qam_100_0.5_bps_lspan8_1amp.csv"   # <-- your second CSV file

# ---- Load both datasets ----
df1 = pd.read_csv(file1)
df2 = pd.read_csv(file2)

# Label datasets so they appear separately in legend
df1["Source"] = "No DPD"
df2["Source"] = "DPD"

# ---- Combine them into one dataframe ----
df = pd.concat([df1, df2], ignore_index=True)

# Scale lspan
df["lspan"] = df["lspan"] / 10

# ---- Create two subplots: DA and NDA ----
fig, axes = plt.subplots(1, 2, figsize=(14, 5), sharey=False)
subplot_map = {True: 0, False: 1}

# ---- Line styles so datasets differ visually ----
style_map = {"No DPD": "-", "DPD": "--"}

for da in [True, False]:
    ax = axes[subplot_map[da]]

    for source in df["Source"].unique():               # No DPD, DPD
        for cpr_mode in df["CPR"].unique():           # bps, ddpll

            subset = df[(df["Data_Aided"] == da) &
                        (df["CPR"] == cpr_mode) &
                        (df["Source"] == source)]

            if len(subset) == 0:
                continue

            subset = subset.sort_values("lspan")

            label = f"{source} - {cpr_mode}"

            ax.semilogy(
                subset["lspan"],
                subset["BER"],
                marker="o",
                linestyle=style_map[source],   # different for each CSV
                label=label
            )

    # Titles
    ax.set_title("DA" if da else "NDA")

    # X label
    ax.set_xlabel("Number of Spans")
    ax.grid(True, which="both", ls="--")

    # Invert x-axis
    ax.invert_xaxis()

    # Y labels
    if da:
        ax.set_ylabel("BER (zoomed)")
    else:
        ax.set_ylabel("BER (NDA scale)")

    # Zoom DA Y-axis
    if da:
        da_vals = df[df["Data_Aided"] == True]["BER"]
        ymin = da_vals.min() * 0.8
        ymax = da_vals.max() * 1.2
        ax.set_ylim([ymin, ymax])

# Legend (place on both or just one)
axes[0].legend(title="Dataset + CPR Mode")
axes[1].legend(title="Dataset + CPR Mode")

plt.suptitle("BER vs lspan — Two CSV Comparison (DA zoomed, NDA full scale)", fontsize=14)
plt.tight_layout()
plt.show()
```
