# 16 QAM No DPD and DPD trained with BPS 100kHz at 0.5 mzm scale

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load both result files
df_no = pd.read_csv("results_No_DPD_16qam_100_mzmchange.csv")
df_dpd = pd.read_csv("results_16qam_100_0.5_DPD_bps.csv")

# Add a label so we know which data is which
df_no["DPD"] = "No DPD"
df_dpd["DPD"] = "DPD_bps"

# Merge into one DataFrame
df = pd.concat([df_no, df_dpd], ignore_index=True)
```

```python
plt.figure(figsize=(10,6))

for dpd_option in df["DPD"].unique():              # "No DPD", "DPD"
    for da in [True, False]:                       # True = DA, False = NDA
        for cpr_mode in df["CPR"].unique():        # "bps", "ddpll"
            
            subset = df[(df["DPD"] == dpd_option) &
                        (df["Data_Aided"] == da) &
                        (df["CPR"] == cpr_mode)]
            
            if len(subset) == 0:
                continue
            
            label = f"{dpd_option} | {'DA' if da else 'NDA'} | {cpr_mode}"
            
            plt.semilogy(
                subset["mzmScale"], 
                subset["BER"], 
                marker='o', 
                linestyle='-', 
                label=label
            )

plt.xlabel("MZM Scale")
plt.ylabel("BER")
plt.title("BER vs MZM Scale (DPD vs No-DPD, DA and NDA)")
plt.grid(True, which='both', ls='--')
plt.legend(fontsize=8)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Read the CSV file directly (no use of 'results')
#df = pd.read_csv("results_No_DPD_lspan_experiment_16qam.csv")
df = pd.read_csv("results_noDPD_lspan.csv")

# Divide lspan by 10
df["lspan"] = 80/df["lspan"]

# Two subplots: DA (left), NDA (right)
fig, axes = plt.subplots(1, 2, figsize=(14, 5), sharey=False)

subplot_map = {True: 0, False: 1}

for da in [True, False]:
    ax = axes[subplot_map[da]]

    for cpr_mode in df["CPR"].unique():

        subset = df[(df["Data_Aided"] == da) &
                    (df["CPR"] == cpr_mode)]

        if len(subset) == 0:
            continue

        # Sort so the plot doesn't zig-zag
        subset = subset.sort_values("lspan")

        label = f"{cpr_mode}"

        ax.semilogy(
            subset["lspan"],
            subset["BER"],
            marker="o",
            linestyle="-",
            label=label
        )

    # Titles
    ax.set_title("DA" if da else "NDA")

    # X label
    ax.set_xlabel("Number of Spans")
    ax.grid(True, which='both', ls='--')

    # Invert x-axis (largest span → smallest)
    #ax.invert_xaxis()

    # Separate Y-axis titles
    if da:
        ax.set_ylabel("BER (zoomed)")
    else:
        ax.set_ylabel("BER (NDA scale)")

    # Zoom DA Y-axis
    if da:
        da_vals = df[df["Data_Aided"] == True]["BER"]
        ymin = da_vals.min() * 0.8
        ymax = da_vals.max() * 1.2
        ax.set_ylim([ymin, ymax])

# Put legend only on NDA subplot
axes[0].legend(title="CPR Mode")
axes[1].legend(title="CPR Mode")

plt.suptitle("BER vs lspan (DA zoomed, NDA separate y-axis)", fontsize=14)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import os
import re

# ---------------------------------------------------------
# 1. CSV files
# ---------------------------------------------------------
csv_files = [
 
    "results_DPD_16qam_3.csv",
    "results_noDPD_16qam_3.csv"
    "results_DPD_64qam_3.csv",
    "results_noDPD_64qam_3.csv",
    
]

# ---------------------------------------------------------
# 2. Load CSVs + extract DPD state from filename
# ---------------------------------------------------------
all_data = []

for f in csv_files:
    df = pd.read_csv(f)

    # Extract "DPD" or "noDPD" from the filename
    match = re.search(r"(DPD|noDPD)", f, re.IGNORECASE)
    dpd_state = match.group(1) if match else "Unknown"

    df["DPD_state"] = dpd_state
    all_data.append(df)

df_all = pd.concat(all_data, ignore_index=True)

print(df_all.head())

# ---------------------------------------------------------
# 3. CONNECT LINES: group by DPD_state + CPR + DA
# ---------------------------------------------------------
plt.figure(figsize=(10, 7))

for (dpd_state, data_aided, cpr_mode), subset in df_all.groupby(["DPD_state", "Data_Aided", "CPR"]):

    subset = subset.sort_values("Modulation")  # ensure 16 → 64 → 256

    label = f"{dpd_state} (DA={data_aided}, {cpr_mode})"

    plt.semilogy(
        subset["Modulation"],
        subset["BER"],
        marker='o',
        linewidth=3,          # thicker line to highlight connection
        label=label
    )

plt.xticks([16, 64, 256], ["16", "64", "256"])
plt.grid(True, which='both', linestyle='--')
plt.xlabel("Modulation Order (M)")
plt.ylabel("BER")
plt.title("BER vs Modulation — DPD vs noDPD (Connected Lines)")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import os
import re

# ---------------------------------------------------------
# 1. CSV files
# ---------------------------------------------------------
csv_files = [
    "results_DPD_64qam_3.csv",
    "results_noDPD_64qam_3.csv",
    "results_DPD_16qam_3.csv",
    "results_noDPD_16qam_3.csv",
]

# ---------------------------------------------------------
# 2. Load CSVs + extract DPD state from filename
# ---------------------------------------------------------
all_data = []

for f in csv_files:
    df = pd.read_csv(f)

    # Extract "DPD" or "noDPD" from the filename
    match = re.search(r"(DPD|noDPD)", f, re.IGNORECASE)
    dpd_state = match.group(1) if match else "Unknown"

    df["DPD_state"] = dpd_state
    all_data.append(df)

df_all = pd.concat(all_data, ignore_index=True)

# ---------------------------------------------------------
# ✅ REMOVE DDPLL (keep only BPS)
# ---------------------------------------------------------
df_all = df_all[df_all["CPR"].str.lower() == "bps"]

print(df_all.head())

# ---------------------------------------------------------
# 3. CONNECT LINES: group by DPD_state + DA
# ---------------------------------------------------------
plt.figure(figsize=(10, 7))

for (dpd_state, data_aided), subset in df_all.groupby(["DPD_state", "Data_Aided"]):

    subset = subset.sort_values("Modulation")

    label = f"{dpd_state} (DA={data_aided}, BPS)"

    plt.semilogy(
        subset["Modulation"],
        subset["BER"],
        marker='o',
        linewidth=3,
        label=label
    )

plt.xticks([16, 64, 256], ["16", "64", "256"])
plt.grid(True, which='both', linestyle='--')

plt.xlabel("Modulation Order (M)")
plt.ylabel("BER")
plt.title("BER vs Modulation — DPD vs noDPD (BPS Only)")

plt.legend()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# ---- CSV FILES TO COMPARE ----
files = [
    ("results_noDPD_lspan.csv", "No DPD"),
    ("results_DPD_NN_lspan_trained_1.csv", "DPD NN"),
    ("results_DPD_MP_lspan_trained_1.csv", "DPD MP"),
    ("results_DPD_WH_lspan_trained_1.csv", "DPD WH"),
]

# ---- Load and label datasets ----
dfs = []
for file, label in files:
    df = pd.read_csv(file)
    df["Source"] = label
    dfs.append(df)

# ---- Combine all ----
df = pd.concat(dfs, ignore_index=True)

# Scale lspan
df["lspan"] = 80/df["lspan"]

# ✅ ---- Keep ONLY DA ----
df = df[df["Data_Aided"] == True]

# ---- Plot ----
plt.figure(figsize=(7, 5))

# Line styles
linestyles = ["-", "--", "-.", ":"]
style_map = {
    source: linestyles[i % len(linestyles)]
    for i, source in enumerate(df["Source"].unique())
}

for source in df["Source"].unique():
    for cpr_mode in df["CPR"].unique():

        subset = df[
            (df["CPR"] == cpr_mode) &
            (df["Source"] == source)
        ]

        if subset.empty:
            continue

        subset = subset.sort_values("lspan")

        label = f"{source} - {cpr_mode}"

        plt.semilogy(
            subset["lspan"],
            subset["BER"],
            marker="o",
            linestyle=style_map[source],
            label=label
        )

# Labels and styling
plt.title("BER vs Lspan (Data Aided Only)")
plt.xlabel("Number of Spans")
plt.ylabel("BER")
plt.grid(True, which="both", ls="--")
#plt.gca().invert_xaxis()

# Optional: zoom like before
ymin = df["BER"].min() * 0.8
ymax = df["BER"].max() * 1.2
plt.ylim([ymin, ymax])

plt.legend()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# ---- CSV FILES TO COMPARE ----
files = [
    ("results_noDPD_lspan.csv", "No DPD"),
    ("results_DPD_NN_lspan_trained_1.csv", "DPD NN"),
    ("results_DPD_MP_lspan_trained_1.csv", "DPD MP"),
    ("results_DPD_WH_lspan_trained_1.csv", "DPD WH"),
]

# ---- Load and label datasets ----
dfs = []
for file, label in files:
    df = pd.read_csv(file)
    df["Source"] = label
    dfs.append(df)

# ---- Combine all ----
df = pd.concat(dfs, ignore_index=True)

# Scale lspan
df["lspan"] = 80 / df["lspan"]

# Keep ONLY Data-Aided
df = df[df["Data_Aided"] == True]

# ---- ✅ Color map (your requirement) ----
color_map = {
    "No DPD": "blue",
    "DPD NN": "red",
    "DPD MP": "orange",
    "DPD WH": "green",
}

# ---- Plot ----
plt.figure(figsize=(7, 5))

for source in df["Source"].unique():
    for cpr_mode in df["CPR"].unique():

        subset = df[
            (df["CPR"] == cpr_mode) &
            (df["Source"] == source)
        ]

        if subset.empty:
            continue

        subset = subset.sort_values("lspan")

        label = f"{source}"

        plt.semilogy(
            subset["lspan"],
            subset["BER"],
            marker="o",
            linestyle="-",                     # ✅ all solid lines
            color=color_map.get(source, "black"),  # ✅ fixed colors
            label=label
        )

# Labels and styling
plt.title("BER vs Lspan (Data Aided BPS)")
plt.xlabel("Number of Spans")
plt.ylabel("BER")
plt.grid(True, which="both", ls="--")

# Axis limits
ymin = df["BER"].min() * 0.8
ymax = df["BER"].max() * 1.2
plt.ylim([ymin, ymax])

plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Pivot table
table = df.pivot_table(
    index="lspan",
    columns="Source",
    values="BER"
)

table = table.sort_index()

# Safe renaming
table = table.rename(columns={
    "No DPD": "No DPD",
    "DPD NN": "NN-DPD",
    "DPD MP": "MP-DPD",
    "DPD WH": "WH-DPD"
})

# ✅ Modern formatting
table_formatted = table.map(lambda x: f"{x:.2e}" if pd.notnull(x) else "-")

print("\n=== Publication-Ready BER Table ===\n")
print(table_formatted.to_string())

# ✅ LaTeX export
print("\n=== LaTeX Table ===\n")
print(table_formatted.to_latex(
    caption="BER vs Number of Spans for Different DPD Methods",
    label="tab:ber_spans",
    column_format="c|cccc"
))
```
